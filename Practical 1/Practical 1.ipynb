```python
# ---------------------------------------------------------
# Perceptron for Financial Market Direction Prediction
#
# Objective:
# Predict whether a stock is expected to have a
# positive or negative return on the next trading day.
#
# Note:
# This is a simple educational example using a synthetic
# dataset to demonstrate how a perceptron learns.
# ---------------------------------------------------------

import numpy as np


# ---------------------------------------------------------
# Step Activation Function
#
# The perceptron calculates a weighted sum of all features.
# If the value is greater than or equal to zero,
# it predicts Class 1.
#
# Otherwise it predicts Class 0.
#
# Output:
# 1 -> Expected Positive Return
# 0 -> Expected Negative Return
# ---------------------------------------------------------

def step_function(x):
    return 1 if x >= 0 else 0


# ---------------------------------------------------------
# Perceptron
#
# A perceptron is the simplest Artificial Neural Network.
#
# It learns by adjusting weights so that its predictions
# become closer to the actual historical outcomes.
#
# It is mainly used to understand the fundamentals of
# supervised learning and neural networks.
# ---------------------------------------------------------

class Perceptron:

    def __init__(self, input_size, learning_rate=0.01):

        # Initialize weights with zero

        self.weights = np.zeros(input_size)

        # Bias term

        self.bias = 0

        # Learning Rate

        self.lr = learning_rate


    # -----------------------------------------------------
    # Prediction
    #
    # z = Weighted Sum + Bias
    #
    # The Step Function converts the score into
    # either 0 or 1.
    # -----------------------------------------------------

    def predict(self, x):

        z = np.dot(x, self.weights) + self.bias

        return step_function(z)


    # -----------------------------------------------------
    # Training
    #
    # Steps:
    #
    # 1. Predict
    # 2. Compare prediction with actual label
    # 3. Calculate error
    # 4. Update weights
    #
    # The perceptron gradually learns which features
    # are important for predicting market direction.
    # -----------------------------------------------------

    def train(self, X, y, epochs=25):

        for epoch in range(epochs):

            for xi, target in zip(X, y):

                prediction = self.predict(xi)

                error = target - prediction

                self.weights += self.lr * error * xi

                self.bias += self.lr * error


# ---------------------------------------------------------
# Financial Dataset
#
# Features:
#
# Daily Return (%)
# -> Percentage change in stock price today.
#
# RSI
# -> Relative Strength Index
# -> Measures market momentum.
#
# Volume Change (%)
# -> Percentage increase/decrease in trading volume.
#
#
# Labels
#
# 1 -> Tomorrow's return was positive.
#
# 0 -> Tomorrow's return was negative.
#
# IMPORTANT:
#
# Daily Return, RSI and Volume DO NOT determine tomorrow's
# return.
#
# They are simply historical indicators.
#
# The perceptron tries to discover statistical
# relationships between today's indicators and
# tomorrow's historical outcome.
#
# In real quantitative finance,
# these labels are created using historical data.
#
# Example:
#
# Today's Features --------> Tomorrow Return
#
# [1.2,63,15] -------------> +2.3%
#
# Since tomorrow's return was positive,
#
# Label = 1
# ---------------------------------------------------------

X = np.array([

    [1.2,63,15],
    [-0.8,35,-22],
    [2.1,72,31],
    [-1.5,28,-12],
    [0.7,58,10],
    [-0.5,40,-8]

])


y = np.array([

    1,
    0,
    1,
    0,
    1,
    0

])


# ---------------------------------------------------------
# Train Model
#
# The perceptron repeatedly adjusts its weights
# over multiple epochs until the prediction error
# becomes smaller.
# ---------------------------------------------------------

model = Perceptron(input_size=3)

model.train(X,y,epochs=25)


# ---------------------------------------------------------
# Predictions
#
# If prediction = 1
#
# Expected Positive Return
#
# If prediction = 0
#
# Expected Negative Return
# ---------------------------------------------------------

print("Predictions\n")

for sample in X:

    prediction = model.predict(sample)

    print(f"Features : {sample}")

    if prediction == 1:

        print("Prediction : Expected Positive Return\n")

    else:

        print("Prediction : Expected Negative Return\n")


# ---------------------------------------------------------
# Limitations
#
# 1. Can only learn linear relationships.
#
# 2. Financial markets are highly non-linear.
#
# 3. Ignores historical sequence information.
#
# 4. Cannot capture complex interactions between
#    multiple financial indicators.
#
# 5. Real financial data contains significant noise.
# ---------------------------------------------------------


# ---------------------------------------------------------
# Better Alternatives
#
# Logistic Regression
# -> Linear probabilistic classifier.
#
# Random Forest
# -> Ensemble of decision trees.
#
# XGBoost / LightGBM
# -> Industry standard models for structured
#    financial datasets.
#
# Multi-Layer Perceptron (MLP)
# -> Learns non-linear relationships.
#
# LSTM
# -> Designed for financial time-series prediction.
#
# Transformer Models
# -> Capture long-range temporal dependencies.
#
# Reinforcement Learning
# -> Learns trading strategies instead of
#    simply predicting market direction.
# ---------------------------------------------------------
```
