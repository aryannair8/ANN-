## Practical-2 Implement a single neuron model manually using Python.

```python
# ==========================================================
# Single Neuron Model (Manual Implementation using Python)
# Finance Application: Loan Approval Prediction
# ==========================================================

# ----------------------------------------------------------
# 1. Importing Library
# ----------------------------------------------------------
import numpy as np
# NumPy is used for numerical computations, arrays, and matrix operations.

# ----------------------------------------------------------
# 2. Defining the Dataset
# ----------------------------------------------------------
# Features:
# Column 1 -> Monthly Income (normalized)
# Column 2 -> Credit Score (normalized)

X = np.array([
    [0.20, 0.30],   # Low income, poor credit
    [0.40, 0.35],   # Medium income, poor credit
    [0.60, 0.70],   # Good income, good credit
    [0.80, 0.90],   # High income, excellent credit
    [0.90, 0.80],   # High income, good credit
    [0.30, 0.60]    # Low income, average credit
])

# Target Output
# 0 = Loan Rejected
# 1 = Loan Approved

y = np.array([
    [0],
    [0],
    [1],
    [1],
    [1],
    [0]
])

# ----------------------------------------------------------
# 3. Initializing Weights and Bias
# ----------------------------------------------------------

# Fixes the random values so that the output remains the same
# every time the program is executed.
np.random.seed(42)

# Initialize two random weights (one for each input feature)
weights = np.random.rand(2, 1)

# Initialize one random bias value
bias = np.random.rand(1)

# ----------------------------------------------------------
# 4. Sigmoid Activation Function
# ----------------------------------------------------------

def sigmoid(x):
    """
    Converts any input value into a probability between 0 and 1.
    Used for binary classification.
    """
    return 1 / (1 + np.exp(-x))

# ----------------------------------------------------------
# 5. Predictions BEFORE Training
# ----------------------------------------------------------

print("Predictions BEFORE Training:\n")

# Weighted Sum:
# z = (X × W) + b
linear = np.dot(X, weights) + bias

# Apply Sigmoid Activation
predictions = sigmoid(linear)

# Display predictions before learning
for i in range(len(X)):
    print(f"Input: {X[i]} -> Prediction: {predictions[i][0]:.4f}")

# ----------------------------------------------------------
# 6. Training the Single Neuron
# ----------------------------------------------------------

learning_rate = 0.5   # Controls how much weights change
epochs = 5000         # Number of training iterations

for epoch in range(epochs):

    # -----------------------------
    # Forward Propagation
    # -----------------------------

    # Compute weighted sum
    linear = np.dot(X, weights) + bias

    # Apply activation function
    output = sigmoid(linear)

    # -----------------------------
    # Error Calculation
    # -----------------------------

    # Difference between actual and predicted output
    error = y - output

    # -----------------------------
    # Gradient Calculation
    # -----------------------------

    # Derivative of sigmoid multiplied by error
    gradient = error * output * (1 - output)

    # -----------------------------
    # Updating Weights
    # -----------------------------

    # Adjust weights to reduce prediction error
    weights += learning_rate * np.dot(X.T, gradient)

    # -----------------------------
    # Updating Bias
    # -----------------------------

    # Adjust bias
    bias += learning_rate * np.sum(gradient)

# ----------------------------------------------------------
# 7. Predictions AFTER Training
# ----------------------------------------------------------

print("\nPredictions AFTER Training:\n")

# Compute final weighted sum
linear = np.dot(X, weights) + bias

# Generate final predictions
predictions = sigmoid(linear)

for i in range(len(X)):

    # Convert probability into class label
    predicted_class = 1 if predictions[i] >= 0.5 else 0

    print(
        f"Input: {X[i]} "
        f"-> Probability: {predictions[i][0]:.4f} "
        f"-> Class: {predicted_class}"
    )

# ----------------------------------------------------------
# 8. Display Learned Parameters
# ----------------------------------------------------------

print("\nLearned Weights:")
print(weights)

print("\nLearned Bias:")
print(bias)

# ----------------------------------------------------------
# Conclusion
# ----------------------------------------------------------
# Initially, the neuron makes random predictions because the
# weights and bias are randomly initialized.
#
# During training, the neuron repeatedly updates its weights
# and bias to minimize the prediction error.
#
# After sufficient training, the neuron learns to classify
# loan applications as Approved (1) or Rejected (0).
```
